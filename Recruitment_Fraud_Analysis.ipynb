### **Replacing null values of categorical columns**

```python
df['employment_type'] = df['employment_type'].fillna('Unknown')
df['required_experience'] = df['required_experience'].fillna('Not Specifed') 
df['required_education'] = df['required_education'].fillna('Not Specifed')
df['function'] = df['function'].fillna('Unknown Function')
df['industry'] = df['industry'].fillna('Unknown Industry')
df['department'] = df['department'].fillna('Unknown Department')
```

```python
df['location'].str.contains('US', na=False).sum()
```

```python
((df['location'].str.contains('US', na=False).sum()/len (df))*100).round(2) 
```

```python
df['salary_range'] = df['salary_range'].fillna('Not Specified')
```

```python
df['location'] = df['location'].fillna('Unknown Location')
df['company_profile'] = df['company_profile'].fillna('No Company Profile Provided') 
df['description'] = df['description'].fillna('Not Available')
df['requirements'] = df['requirements'].fillna('Not Specified')
df['benefits'] = df['benefits'].fillna('Not Specified') 
```

### **Understand target variables**

```python
total_jobs = len(df)
real_jobs = (df['fraudulent'] == 0).sum() 
fake_jobs = (df['fraudulent'] == 1).sum() 
fraud_percentage = (fake_jobs / total_jobs)*100
```

```python
print("Total Jobs:", total_jobs)
print("Real Jobs:", real_jobs)
print("Fake Jobs:", fake_jobs)
print("Fraud Percentage:", round(fraud_percentage,2), "%")
```
